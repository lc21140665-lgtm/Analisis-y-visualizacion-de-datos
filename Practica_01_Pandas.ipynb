```python
df_limpio = df.copy()
```

```python
df_limpio.shape
```

```python
df_limpio = df_limpio.drop_duplicates()
```

```python
df_limpio.shape
```

¿Disminuyó el número de registros? Disminuyó

```python
df_limpio["Cliente"] = df_limpio["Cliente"].str.strip()
```

```python
df_limpio["Vendedor"] = df_limpio["Vendedor"].str.title()
```

```python
df_limpio["Ciudad"] = df_limpio["Ciudad"].replace({
    "QUERETARO":"Querétaro",
    "Queretaro":"Querétaro",
    "queretaro":"Querétaro",
    "Qro":"Querétaro"
})
```

¿Qué ocurrió con las variantes?
Todas las variantes (Queretaro, queretaro, QUERETARO y Qro) fueron reemplazadas por un único nombre: Querétaro, dejando un solo valor consistente para la ciudad.

```python
df_limpio["Producto"] = df_limpio["Producto"].str.title()
```

```python
df_limpio[df_limpio["Cantidad"] < 0]
```

```python
df_limpio = df_limpio[df_limpio["Cantidad"] >= 0]
```

¿Por qué decidimos eliminarlos y no convertirlos en positivos?
Porque una cantidad negativa representa un error de captura. Convertirla a positiva modificaría la información original y podría generar resultados incorrectos. Por ello, es mejor eliminar esos registros, de acuerdo con las reglas del negocio.

```python
df_limpio = df_limpio[df_limpio["Descuento (%)"] <= 100]
```

```python
df_limpio["IVA (%)"] = df_limpio["IVA (%)"].fillna(16)
```

```python
df_limpio.duplicated().sum()
```

```python
df_limpio["Ciudad"].value_counts()
```

PRACTICA **6**

```python
df_limpio
```

```python
df_limpio["Subtotal"] = df_limpio["Cantidad"] * df_limpio["Precio Unitario"]
```

```python
df_limpio.head()
```

```python
df_limpio["Monto Descuento"] = (
    df_limpio["Subtotal"] *
    df_limpio["Descuento (%)"]/100
)
```

```python
df_limpio["Subtotal Neto"] = (
    df_limpio["Subtotal"] -
    df_limpio["Monto Descuento"]
)
```

```python
df_limpio["IVA"] = (
    df_limpio["Subtotal Neto"] *
    df_limpio["IVA (%)"]/100
)
```

```python
df_limpio["Total"] = (
    df_limpio["Subtotal Neto"] +
    df_limpio["IVA"]
)
```

```python
df_limpio["Costo Total"] = (
    df_limpio["Cantidad"] *
    df_limpio["Costo Unitario"]
)
```

```python
df_limpio["Utilidad"] = (
    df_limpio["Subtotal Neto"] -
    df_limpio["Costo Total"]
)
```

```python
df_limpio[
[
"Cantidad",
"Precio Unitario",
"Subtotal",
"Monto Descuento",
"Subtotal Neto",
"IVA",
"Total",
"Costo Total",
"Utilidad"
]
].head()
```

```python
df_limpio = df_limpio.round(2)
```
